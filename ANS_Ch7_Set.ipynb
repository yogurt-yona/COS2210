ในภาคการศึกษาล่าสุด สาขาวิทยาการคอมพิวเตอร์ได้เปิดสอนวิชาเลือกด้านการเขียนโปรแกรม 3 วิชา ได้แก่
* Python, Java, และ C++ โดยมีรายชื่อนักศึกษาที่ลงทะเบียนในแต่ละวิชาดังนี้
วิชา Python มีผู้ลงทะเบียน 6 คน ได้แก่: SUPANAT, TEERAPON, KAWINWAT, KIDSAKORN, PHON, SURACHAD

* วิชา Java มีผู้ลงทะเบียน 6 คน ได้แก่: PAWARIS, SUPANAT, PONGSAKORN, RATTANACHAT, PHON, SURACHAD

* วิชา C++ มีผู้ลงทะเบียน 6 คน ได้แก่: SUPANAT, TEERAPON, KAWINWAT, RATTANACHAT, SURACHAD, MATCHAKAN


จงเขียนโปรแกรมเพื่อหาจำนวนและรายชื่อของนักศึกษาที่ลงทะเบียนเรียนต่อไปนี้

1.   ครบทั้ง 3 วิชา
2.   เฉพาะวิชา Python เพียงวิชาเดียว
3. เฉพาะวิชา Java เพียงวิชาเดียว
4. เฉพาะวิชา C++ เพียงวิชาเดียว
5. วิชา Python และ Java เท่านั้น (แต่ไม่ได้เรียน C++)
6. วิชา Python และ C++ เท่านั้น (แต่ไม่ได้เรียน Java)
7. วิชา Java และ C++ เท่านั้น (แต่ไม่ได้เรียน Python)
8. อย่างน้อย 2 วิชาขึ้นไป


```python
# 1. กำหนดเซตของนักศึกษาในแต่ละวิชา
python = {"SUPANAT", "TEERAPON", "KAWINWAT", "KIDSAKORN", "PHON", "SURACHAD"}
java = {"PAWARIS", "SUPANAT", "PONGSAKORN", "RATTANACHAT", "PHON", "SURACHAD"}
cpp = {"SUPANAT", "TEERAPON", "KAWINWAT", "RATTANACHAT", "SURACHAD", "MATCHAKAN"}

# 2. คำนวณหาผลลัพธ์ตามเงื่อนไขเซต
ans1 = python & java & cpp
ans2 = python - java - cpp
ans3 = java - python - cpp
ans4 = cpp - python - java
ans5 = (python & java) - cpp
ans6 = (python & cpp) - java
ans7 = (java & cpp) - python
ans8 = (python & java) | (python & cpp) | (java & cpp)

# 3. แสดงผลลัพธ์
print(f"1. ลงเรียนครบทั้ง 3 วิชา ({len(ans1)} คน): {', '.join(ans1)}")
print(f"2. เฉพาะวิชา Python เพียงวิชาเดียว ({len(ans2)} คน): {', '.join(ans2)}")
print(f"3. เฉพาะวิชา Java เพียงวิชาเดียว ({len(ans3)} คน): {', '.join(ans3)}")
print(f"4. เฉพาะวิชา C++ เพียงวิชาเดียว ({len(ans4)} คน): {', '.join(ans4)}")
print(f"5. วิชา Python และ Java เท่านั้น ({len(ans5)} คน): {', '.join(ans5)}")
print(f"6. วิชา Python และ C++ เท่านั้น ({len(ans6)} คน): {', '.join(ans6)}")
print(f"7. วิชา Java และ C++ เท่านั้น ({len(ans7)} คน): {', '.join(ans7)}")
print(f"8. เรียนอย่างน้อย 2 วิชาขึ้นไป ({len(ans8)} คน): {', '.join(ans8)}")
```
