Ex3. เขียนโปรแกรมรับค่าเลขจำนวนเต็ม 3 ตัวทีละบรรทัด (ทีละตัว) จากนั้นให้แสดงผลรวม (Summation) บนหน้าจอตามรูปแบบที่กำหนด


**รูปแบบเอาต์พุต:**
```
<1stInteger> + <2ndInteger> + <3rdInteger> = <summation>
```

````python
num1 = int(input())
num2 = int(input())
num3 = int(input())
summation = num1 + num2 + num3
print(f"{num1} + {num2} + {num3} = {summation}")
````

Ex4. ขียนโปรแกรมรับค่าเลขจำนวนเต็ม 3 ตัวพร้อมกันในบรรทัดเดียว โดยคั่นด้วยเครื่องหมายจุลภาค (,) จากนั้นให้แสดงผลรวมบนหน้าจอตามรูปแบบที่กำหนด


**รูปแบบเอาต์พุต:**
```
<1stInteger> + <2ndInteger> + <3rdInteger> = <summation>
```

````python
num1, num2, num3 = map(int, input().split(","))
summation = num1 + num2 + num3
print(f"{num1} + {num2} + {num3} = {summation}")
````

Ex5. เขียนโปรแกรมรับราคาสินค้าที่รวมภาษีมูลค่าเพิ่ม 7% (Vat) แล้ว จากนั้นคำนวณและแสดงผลเฉพาะมูลค่าของภาษี (Vat value) ออกทางหน้าจอ


**รูปแบบเอาต์พุต:**

```python
Enter price with vat: 347.75<enter>
vat is 22.75 Baht.
```

````python
price_with_vat = float(input("Enter price with vat: "))
vat_value = price_with_vat * (7 / 107)
print(f"vat is {vat_value:.2f} Bath.")
````

Ex7. เขียนโปรแกรมรับช่วงเวลาเป็นวินาที (Duration in seconds) จากนั้นแปลงให้อยู่ในรูปแบบ
```
<hour>:<min>:<sec>
```

Ex.
```python
Enter duration: 3690<enter>
1:1:30
```

````python
total_seconds = int(input("Enter duration: "))
hours = total_seconds // 3600
remaining_seconds = total_seconds % 3600
minutes = remaining_seconds // 60
seconds = remaining_seconds % 60
print(f"{hours}:{minutes}:{seconds}")
````
