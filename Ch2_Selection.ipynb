2. จงเขียนโปรแกรมคำนวณราคาห้องพักของโรงแรมตามเงื่อนไขในตารางที่กำหนดให้ โดยอันดับแรกให้รับข้อมูลประเภทห้องพัก (Room type) จากผู้ใช้ หากประเภทห้องพักถูกต้อง ให้โปรแกรมถามจำนวนคืนที่ต้องการเข้าพัก (Number of nights to stay) แต่หากประเภทห้องพักไม่ถูกต้อง ให้แสดงข้อความเตือนว่า "INVALID CHOICE"ในกรณีที่ประเภทห้องพักถูกต้อง แต่ผู้ใช้ระบุจำนวนคืนเข้าพัก $\le 0$ ให้โปรแกรมแสดงข้อความเตือนว่า "INVALID NIGHT"

| ----- | Room type | Price per night | |
| ----- | ------ | -------- | ---- |
|       |        | 1 - 2 nights | > 2 nights |
| Deluxe  | D, d   | 3500 Baht | 3000 Baht |
| Grand   | G, g   | 4000 Baht | 4000 Baht |

Ex.
```
Room type: x<enter>
INVALID CHOICE
```
```
Room type: d<enter>
Nights to stay: 0<enter>
INVALID NIGHT
```
```
Room type: d<enter>
Nights to stay: 5<enter>
Total price is 15000 baht.
```

````python
room_type = input("Room type: " )
if room_type == 'D' or room_type == 'd' or room_type == 'G' or room_type =='g':
  nights = int(input("Nights to stay: "))
  if nights <= 0:
    print("INVALID NIGHT")
  else:
    if room_type == 'D' or room_type == 'd':
      if nights <=2:
        price_per_night = 3500
      else:
        price_per_night =3000
    elif room_type == 'G' or room_type == 'g':
      price_per_night =4000
    total_price = price_per_night * nights
    print(f"Total price is {total_price} bath.")
else:
  print("INVAID CHOICE")
````

4. **โปรแกรมคำนวณภาษี (Tax calculator)**

จงเขียนโปรแกรมเพื่อรับข้อมูลเงินเดือน (Salary) และเงินค่าเบี้ยเลี้ยงประจำปีเพิ่มเติม (Extra annual allowance) จากผู้ใช้ จากนั้นคำนวณและแสดงผลจำนวนภาษีที่ต้องชำระ โดยอ้างอิงตามเกณฑ์ที่กำหนดในตารางด้านล่างนี้

รายได้สุทธิ (Net income) | อัตราภาษี (Tax rate)
-------------| ------
0 - 150,000  | ได้รับการยกเว้น (Waived)
มากกว่า 150,000 - 300,000 | 5%
มากกว่า 300,000 - 500,000 | 10%
มากกว่า 500,000 | 15%



````python
salary = float(input("Enter monthly salary: "))
allowance = float(input("Enter exter annual allowance:"))
net_income = (salary * 12) + allowance
if net_income <= 150000:
  tex_rate = 0.00
elif net_income <=300000:
  tex_rate = 0.05
elif net_income <= 500000:
  tex_rate = 0.10
else:
  tex_rate = 0.15
tax_amount = net_income * tex_rate
print(f"Net income: {net_income:,.2f} bath")
print(f"Tax rate: {int(tex_rate * 100)}%")
print(f"Tax amount: {tax_amount:,.2f} bath")
````

5. จงเขียนโปรแกรมเพื่อรับค่าตัวเลขสามจำนวน ได้แก่ $a$, $b$, และ $c$ โดยป้อนข้อมูลทั้งหมดภายในบรรทัดเดียวกันและคั่นด้วยเครื่องหมายจุลภาค ( , ) จากนั้นให้คำนวณผลลัพธ์โดยใช้สูตรต่อไปนี้  

$-b \pm \dfrac{\sqrt{b^2-4ac}}{2a}$,

และแสดงคำตอบที่ได้ออกทางหน้าจอ โดยมีเงื่อนไขการแสดงผลดังนี้:
หากมีคำตอบเพียงคำตอบเดียว ให้แสดงผลเฉพาะค่า ๆ นั้น
หากค่าที่อยู่ภายในเครื่องหมายรากที่สอง (Square root) มีค่าติดลบ ให้แสดงข้อความเตือนว่า "No real solution"

````python
import math
a,b,c = map(float, input("Enter a, b, c").split(','))
discriminant = b**2 - 4*a*c
if discriminant <0:
  print("No real solution")
elif discriminant ==0:
  x = -b / (2*a)
  print(f"x = {x}")
else:
  x1 = (-b + math.sqrt(discriminant)) / (2*a)
  x2 = (-b - math.sqrt(discriminant)) / (2*a)
  print(f"x1 = {x1}, x2 = {x2}")
````
