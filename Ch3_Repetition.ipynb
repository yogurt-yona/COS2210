1. จงเขียนโปรแกรมรับค่าเงินฝากสะสม (Saving amount / Principle) และอัตราดอกเบี้ย (Interest rate) จากผู้ใช้ จากนั้นให้แสดงผลเงินต้น ดอกเบี้ยที่ได้รับ และเงินรวม ของแต่ละปีต่อเนื่องกันเป็นเวลา 5 ปี (หมายเหตุ: ดอกเบี้ยที่ได้จะถูกนำไปสมทบเป็นเงินต้นของปีถัดไปในลักษณะดอกเบี้ยทบต้นทุกปี)

Ex.
```
Enter principle : 1000<enter> Enter interest rate : 0.5<enter>
year 1 Principle : 1000.0 , Interest : 5.0 , Balance : 1005.0
year 2 Principle : 1005.0 , Interest : 5.025 , Balance : 1010.025
year 3 Principle : 1010.025 , Interest : 5.0501249999999995 , Balance : 1015.075125
year 4 Principle : 1015.075125 , Interest : 5.0753756249999995 , Balance : 1020.150500625
year 5 Principle : 1020.150500625 , Interest : 5.100752503124999 , Balance : 1025.251253128125
```

````python
# 1. รับค่าเงินต้น และอัตราดอกเบี้ย
principle = float(input("Enter principle : "))
interest_rate = float(input("Enter interest rate : "))

# 2. วนลูปคำนวณ 5 ปี (ใช้ range(1, 6) เพื่อให้ปีเริ่มจาก 1 ถึง 5)
for year in range(1, 6):
    # คำนวณดอกเบี้ย
    interest = principle * (interest_rate / 100)

    # คำนวณเงินรวมสิ้นปี
    balance = principle + interest

    # แสดงผลตามรูปแบบในตัวอย่างเป๊ะๆ
    print(f"year {year} Principle : {principle} , Interest : {interest} , Balance : {balance}")

    # อัปเดตเงินต้นของปีถัดให้เท่ากับเงินรวมปีนี้ (ดอกเบี้ยทบต้น)
    principle = balance
````

2. Write a program that recieves an integer, n, from the user, which is the Fibonacci number, then print out the last value of the Fibonacci sequence where each number in the sequence is reated as follows.
* Fibonacci(0) = 0
* Fibonacci(1) = 1
* Fibonacci(n) = Fibonacci(n-1) + Fibonacci(n-2)

Ex.
```
Enter n : 0<enter>
Fibonacci(0) = 0
```
```
Enter n : 10<enter>
Fibonacci( 10 ) = 55
```

````python
# 1. รับค่า n จากผู้ใช้
n = int(input("Enter n : "))

# 2. กรณีพิเศษถ้า n = 0 หรือ n = 1
if n == 0:
    print("Fibonacci(0) = 0")
elif n == 1:
    print("Fibonacci( 1 ) = 1")
else:
    # เริ่มต้นค่า F(0) = 0, F(1) = 1
    a, b = 0, 1

    # วนลูปคำนวณตั้งแต่ลำดับที่ 2 ถึง n
    for _ in range(2, n + 1):
        a, b = b, a + b

    # แสดงผลลัพธ์ตามฟอร์แมตในตัวอย่าง
    print(f"Fibonacci( {n} ) = {b}")
````

3. จงเขียนโปรแกรมรับค่าจำนวนเต็มบวก $n$ จากผู้ใช้ จากนั้นให้ทำการตรวจสอบว่าจำนวนดังกล่าวเป็นจำนวนเฉพาะ (Prime number) หรือไม่

````python
import math

n = int(input("Enter n: "))

if n <= 1:
    is_prime = False
else:
    is_prime = True
    # วนลูปเช็กแค่ถึง sqrt(n)
    for i in range(2, int(math.isqrt(n)) + 1):
        if n % i == 0:
            is_prime = False
            break

if is_prime:
    print(f"{n} is a prime number.")
else:
    print(f"{n} is not a prime number.")
````

4. จงเขียนโปรแกรมเพื่อแสดงผลลัพธ์ของค่าแฟกทอเรียล ตั้งแต่ $1!, 2!, 3!, \dots, 20!$ ต่อเนื่องกันทีละบรรทัด


````python
# กำหนดค่าเริ่มต้นสำหรับเก็บผลคูณแฟกทอเรียล
factorial = 1

# วนลูปตั้งแต่ 1 ถึง 20 (ใช้ range(1, 21))
for i in range(1, 21):
    factorial *= i  # หรือ factorial = factorial * i
    print(f"{i}! = {factorial}")
````

5. จงเขียนโปรแกรมเพื่อแสดงผลลัพธ์ออกทางหน้าจอตามรูปแบบต่อไปนี้

1 + 100 = 101

2 + 99 = 101

3 + 98 = 101

...

48 + 53 = 101

````python
# วนลูปตั้งแต่ 1 ถึง 48 ( range(1, 49) )
for i in range(1, 49):
    j = 101 - i
    print(f"{i} + {j} = {i + j}")
````
