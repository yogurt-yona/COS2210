1. จงเขียนโปรแกรมรับลิสต์ของเลขจำนวนเต็ม 2 ชุด แล้วตรวจสอบว่าลิสต์ทั้งสองนั้นมีลักษณะการเรียงตัวเป็นวงกลมที่เหมือนกันหรือไม่ (Circularly Identical) ให้พิมพ์คำว่า "yes" ถ้าใช่ และพิมพ์คำว่า "no" ถ้าไม่ใช่


ตัวอย่าง

please input values in list1:10 10 0 0 10

please input values in list2:10 10 10 0 0

yes


```python
# 1. รับข้อมูลลิสต์ทั้งสองชุดจากผู้ใช้
list1 = input("please input values in list1:").split()
list2 = input("please input values in list2:").split()

is_identical = False

# 2. ตรวจสอบความยาว และลองหมุนลิสต์เพื่อเปรียบเทียบ
if len(list1) == len(list2):
    for i in range(len(list1)):
        # หมุน list1 ไป i ตำแหน่ง แล้วเช็กว่าตรงกับ list2 หรือไม่
        if list1[i:] + list1[:i] == list2:
            is_identical = True
            break

# 3. แสดงผลลัพธ์
if is_identical:
    print("yes")
else:
    print("no")
```

2. หุ่นยนต์เคลื่อนที่บนระนาบ 2 มิติ โดยเริ่มต้นจากจุดกำเนิด $(0,0)$ หุ่นยนต์สามารถเคลื่อนที่ไปในทิศทาง บน (U), ล่าง (D), ซ้าย (L) และ ขวา (R) ตามคำสั่งที่ป้อนเข้า เช่น U5 D3 L3 R2 (ตัวเลขที่อยู่หลังอักษรบอกทิศทางคือจำนวนก้าว) จงคำนวณหาระยะห่างทางตรงจากจุดเริ่มต้นหลังจากหุ่นยนต์เคลื่อนที่ตามชุดคำสั่งเสร็จสิ้น

ตัวอย่าง

Enter a sequence of move :U 5 D 3 L 3 R 2

2.23606797749979

```python
import math

# 1. รับชุดคำสั่งการเคลื่อนที่
user_input = input("Enter a sequence of move :")

# แยกคำสั่งด้วยช่องว่าง
tokens = user_input.split()

x, y = 0, 0
i = 0

# 2. วนลูปอ่านคำสั่งทิศทางและจำนวนก้าว
while i < len(tokens):
    # กรณีป้อนคำสั่งแบบเว้นวรรคแยกทิศทางกับตัวเลข เช่น U 5 D 3
    if tokens[i].isalpha() and i + 1 < len(tokens) and tokens[i+1].isdigit():
        direction = tokens[i].upper()
        steps = int(tokens[i+1])
        i += 2
    # กรณีป้อนคำสั่งแบบติดกัน เช่น U5 D3
    else:
        direction = tokens[i][0].upper()
        steps = int(tokens[i][1:])
        i += 1

    # อัปเดตพิกัด (x, y)
    if direction == 'U':
        y += steps
    elif direction == 'D':
        y -= steps
    elif direction == 'R':
        x += steps
    elif direction == 'L':
        x -= steps

# 3. คำนวณระยะทางจากจุดเริ่มต้น (0,0)
distance = math.sqrt(x**2 + y**2)

# 4. แสดงผลลัพธ์
print(distance)
```

3. จงเขียนโปรแกรมคำนวณยอดเงินคงเหลือสุทธิ (Net Amount) จากบันทึกรายการธุรกรรมที่รับเข้ามา โดยข้อมูลทั้งหมดจะป้อนเข้ามาในบรรทัดเดียวในรูปแบบ [W|D number] เช่น D 300 D 300 W 200 D 300 (ตัวอักษรอ้างอิง: W หมายถึงการถอนเงิน และ D หมายถึงการฝากเงิน)

ตัวอย่าง

Specify deposit and withdrawal amounts: D 300 D 300 W 200 D 300

Total amount is  700


```python
# 1. รับข้อมูลบันทึกธุรกรรม
user_input = input("Specify deposit and withdrawal amounts: ")

# แยกข้อมูลออกเป็นลิสต์
tokens = user_input.split()

net_amount = 0

# 2. วนลูปทีละ 2 ข้อมูล (ขยับทีละ step = 2)
for i in range(0, len(tokens), 2):
    action = tokens[i].upper()
    amount = int(tokens[i + 1])

    if action == 'D':
        net_amount += amount
    elif action == 'W':
        net_amount -= amount

# 3. แสดงผลลัพธ์ตามฟอร์แมตในตัวอย่าง
print(f"Total amount is  {net_amount}")
```

4. จงเขียนโปรแกรมสำหรับตรวจสอบความถูกต้องของรหัสผ่านที่กำหนด โดยรหัสผ่านที่ยอมรับได้จะต้องมีเงื่อนไขดังนี้: ปรากฏตัวอักษร c อย่างน้อย 1 ตัว และปรากฏตัวเลข 1 ตัวพิมพ์ตรงเป๊ะเพียงแค่ 1 ตัวเท่านั้น

* ตัวอย่างรหัสผ่านที่ผ่านเงื่อนไข: ABcd0123bca$ (พิมพ์คำว่า valid)

* ตัวอย่างรหัสผ่านที่ไม่ผ่านเงื่อนไข: c11 (พิมพ์คำว่า invalid เพราะมีเลข 1 ซ้ำสองตัว)

ตัวอย่าง

Please enter the password you want to check for validity : ABcd0123bca$

The password is valid

```python
# 1. รับค่ารหัสผ่านจากผู้ใช้
password = input("Please enter the password you want to check for validity : ")

# 2. ตรวจสอบเงื่อนไข
# - เงื่อนไขที่ 1: มี 'c' อย่างน้อย 1 ตัว ('c' in password)
# - เงื่อนไขที่ 2: มีเลข '1' เพียงแค่ตัวเดียวเท่านั้น (password.count('1') == 1)
if ('c' in password) and (password.count('1') == 1):
    print("The password is valid")
else:
    print("The password is invalid")
```

5. จงเขียนโปรแกรมรับข้อมูลพิกัดมุมบนซ้าย $(x, y)$, ความกว้าง (width) และความสูง (height) ของรูปสี่เหลี่ยมผืนผ้า 2 รูป เช่น รูปที่ 1: [5, 5, 10, 10] และ รูปที่ 2: [0, 0, 12, 15] จากนั้นให้แสดงผลลัพธ์คำว่า "yes" หากพื้นที่ของรูปสี่เหลี่ยมทั้งสองมีส่วนที่ทับซ้อนหรือตัดกัน (Intersect) และให้พิมพ์คำว่า "no" ในกรณีที่แยกออกจากกันอย่างเด็ดขาด

```python
# 1. รับข้อมูลสี่เหลี่ยมรูปที่ 1 และ 2 (x, y, width, height)
# ตัวอย่างอินพุต: 5 5 10 10
rect1 = list(map(float, input("Enter Rectangle 1 (x y w h): ").split()))
rect2 = list(map(float, input("Enter Rectangle 2 (x y w h): ").split()))

x1, y1, w1, h1 = rect1
x2, y2, w2, h2 = rect2

# 2. เช็กเงื่อนไขการไม่ทับซ้อนกันเด็ดขาด (Separated)
no_overlap = (x1 + w1 <= x2) or (x1 >= x2 + w2) or (y1 + h1 <= y2) or (y1 >= y2 + h2)

# 3. แสดงผลลัพธ์
if not no_overlap:
    print("yes")
else:
    print("no")
```

6. จงรับค่าจำนวนนักเรียนและคะแนนของนักเรียนเพื่อเก็บลงลิสต์ แล้วหาคะแนนต่ำสุด สูงสุด ค่าเฉลี่ย ค่ามัธยฐาน

```python
import statistics

# 1. รับจำนวนนักเรียน
n = int(input("Enter number of students: "))

scores = []
# 2. วนลูปรับคะแนนนักเรียนตามจำนวน n
for i in range(n):
    score = float(input(f"Enter score for student {i+1}: "))
    scores.append(score)

# 3. คำนวณค่าต่างๆ
min_score = min(scores)
max_score = max(scores)
avg_score = sum(scores) / len(scores)

# หาค่ามัธยฐาน (Median) แบบเขียน Logic เอง
sorted_scores = sorted(scores)
mid = len(sorted_scores) // 2

if len(sorted_scores) % 2 == 1:
    median_score = sorted_scores[mid]
else:
    median_score = (sorted_scores[mid - 1] + sorted_scores[mid]) / 2

# 4. แสดงผลลัพธ์
print("--- Results ---")
print(f"Minimum score: {min_score}")
print(f"Maximum score: {max_score}")
print(f"Average score: {avg_score:.2f}")
print(f"Median score: {median_score}")
```
