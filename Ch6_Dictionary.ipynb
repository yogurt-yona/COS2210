1. จงสร้าง `dict2`โดยสลับคู่ข้อมูลระหว่าง Key และ Value จาก `dict1` ด้วยการใช้ dict comprehension

ตัวอย่างเช่น:

ถ้ากำหนดให้ `dict1 = {'a':1, 'b':2, 'c':3}`

ผลลัพธ์ที่ต้องได้คือ
`dict2 = {1:'a', 2:'b', 3:'c'}`.


````python
# 1. กำหนด dict1 ตามโจทย์
dict1 = {'a': 1, 'b': 2, 'c': 3}

# 2. สลับ Key และ Value ด้วย Dict Comprehension
dict2 = {value: key for key, value in dict1.items()}

# 3. แสดงผลลัพธ์
print(f"dict2 = {dict2}")
````

2. จงเขียนโปรแกรมเพื่อรับจำนวนของรถเมล์ จากนั้นในรถเมล์แต่ละคัน ให้แอดมิน (Admin) กรอกหมายเลขรถเมล์และจุดหมายปลายทาง หลังจากนั้น ให้ถามผู้โดยสาร (Passenger) ว่าต้องการไปที่ไหน แล้วทำการแสดงหมายเลขรถเมล์ทั้งหมดที่ผู้โดยสารสามารถนั่งไปได้

ตัวอย่างการทำงาน
```
Enter n: 5
Enter bus no. and destination 1: 8, Happyland
Enter bus no. and destination 2: 145, Mo Chit 2
Enter bus no. and destination 3: 122, Mo Chit 2
Enter bus no. and destination 4: 157, Chatuchak
Enter bus no. and destination 5: 509, Mo Chit 2
Enter your destination: Mo Chit 2
You can get bus no. 145 122 509
```


````python
# 1. รับจำนวนรถเมล์
n = int(input("Enter n: "))

# ใช้ Dictionary เก็บข้อมูล {จุดหมายปลายทาง: [สายรถเมล์1, สายรถเมล์2, ...]}
buses = {}

# 2. แอดมินกรอกข้อมูลรถเมล์แต่ละคัน
for i in range(1, n + 1):
    data = input(f"Enter bus no. and destination {i}: ").split(',')
    bus_no = data[0].strip()
    destination = data[1].strip()

    # ถ้ายังไม่มีจุดหมายนี้ใน dict ให้สร้างลิสต์ใหม่
    if destination not in buses:
        buses[destination] = []

    # เพิ่มสายรถเมล์ลงในจุดหมายนั้นๆ
    buses[destination].append(bus_no)

# 3. ถามผู้โดยสารว่าต้องการไปที่ไหน
target = input("Enter your destination: ").strip()

# 4. ตรวจสอบและแสดงผลลัพธ์
if target in buses:
    available_buses = " ".join(buses[target])
    print(f"You can get bus no. {available_buses}")
else:
    print("No bus found for this destination.")
````

3. จงเขียนโปรแกรมระบบแคชเชียร์ (Cashier Program) เพื่อทำหน้าที่ดังต่อไปนี้

* รับค่าซ้ำๆ โดยให้แอดมินกรอก ชื่อสินค้า, จำนวนที่ซื้อ, และราคาต่อหน่วย ไปเรื่อยๆ จนกว่าผู้ใช้จะพิมพ์คำว่า `Done`
* สร้าง Dictionary ชื่อ `shopping_cart` โดยให้ Key คือชื่อสินค้า และ Value คือจำนวนของสินค้านั้น ๆ
* สร้าง Dictionary ชื่อ `prices` โดยให้ Key คือชื่อสินค้า และ Value คือราคาต่อหน่วยของสินค้านั้น ๆ
* คำนวณราคารวมทั้งหมด ที่ลูกค้าต้องจ่าย

*Note* นค้าชนิดเดียวกันควรจะมีราคาต่อหน่วยเท่ากัน แต่หากผู้ใช้ป้อนราคามาไม่เท่ากัน ให้ใช้ราคาล่าสุดในการคำนวณ

ตัวอย่างการทำงาน

```
Enter item 1: mango, 4, 20.5
Enter item 2: rice pudding, 5, 100
Enter item 3: milk, 2, 30
Enter item 4: mango, 2, 20.5
Enter item 5: Done

You have to pay: 683.0
mango: 123.0
rice: 500.0
milk: 60.0
```

````python
shopping_cart = {}
prices = {}

item_count = 1

while True:
    user_input = input(f"Enter item {item_count}: ").strip()

    # เช็กเงื่อนไขการเลิกลูป
    if user_input.lower() == 'done':
        break

    # แยกข้อมูลด้วยเครื่องหมายจุลภาค (,)
    parts = user_input.split(',')
    item = parts[0].strip()
    qty = int(parts[1].strip())
    price = float(parts[2].strip())

    # อัปเดตจำนวนสินค้าใน shopping_cart (สะสมจำนวน)
    if item in shopping_cart:
        shopping_cart[item] += qty
    else:
        shopping_cart[item] = qty

    # อัปเดตราคาต่อหน่วยใน prices (ทับเป็นราคาล่าสุดเสมอ)
    prices[item] = price

    item_count += 1

# คำนวณราคารวมทั้งหมด
total_pay = 0
items_summary = []

for item, qty in shopping_cart.items():
    item_total = qty * prices[item]
    total_pay += item_total
    items_summary.append((item, item_total))

# แสดงผลลัพธ์
print(f"\nYou have to pay: {total_pay}")
for item, item_total in items_summary:
    print(f"{item}: {item_total}")
````
