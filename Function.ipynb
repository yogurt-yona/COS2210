1. จงเขียน function ชื่อ `sum_digits` ที่รับค่าจำนวนเต็ม (integer) เข้ามา และ return ผลรวมของตัวเลขในแต่ละหลัก

ตัวอย่าง

```python
print(sum_digits(1234)) # output: 10
```

````python
def sum_digits(n):
    total = 0               # 1. สร้างตัวแปรมาคอยบวกสะสมผลรวม
    while n>0:              # 2. ทำวนลูปตราบใดที่ n ยังมากกว่า 0
        last = n % 10       # 3. เอา % 10 เพื่อ "ดึงเลขหลักหน่วย" ออกมา
        total += last       # 4. เอาเลขหลักหน่วยที่ได้ไปบวกสะสมเข้า total
        n //= 10            # 5. เอา // 10 เพื่อ "ตัดเลขหลักหน่วยทิ้งไป"
    return total            # 6. พอ n เหลือ 0 (หมดทุกหลัก) ก็ส่งคืนผลรวม
num=int(input("Enter integer: "))
result=sum_digits(num)
print(result)
````

3. จงเขียน function ชื่อ `check_date` ที่รับค่าจำนวนเต็ม 3 ค่า ได้แก่ `m` (เดือน), `d` (วันที่), และ `y` (ปี ค.ศ.) แล้ว return ค่า Boolean ว่าวันที่ที่ระบุนั้นเป็นวันที่ถูกต้องหรือไม่ ทั้งนี้ ให้สังเกตว่า Leap year (ปีอธิกสุรทินที่มี 29 วันในเดือนกุมภาพันธ์) คือปีที่หารด้วย 4 ลงตัวแต่หารด้วย 100 ไม่ลงตัว หรือเป็นปีที่หารด้วย 400 ลงตัว

````python
def check_date(m, d, y):
    # 1. เช็กความถูกต้องของเดือนเบื้องต้น
    if m < 1 or m > 12:
        return False
    # 2. เช็กว่าเป็นปีอธิกสุรทิน (Leap Year) หรือไม่
    is_leap_year = (y % 400 == 0) or (y % 4 == 0 and y % 100 != 0)
    # 3. กำหนดจำนวนวันสูงสุดในแต่ละเดือน
    if m in [1, 3, 5, 7, 8, 10, 12]:
        max_days = 31
    elif m in [4, 6, 9, 11]:
        max_days = 30
    elif m == 2:
        if is_leap_year:
            max_days = 29
        else:
            max_days = 28
    # 4. เช็กว่าวันที่ (d) อยู่ในช่วง 1 ถึง max_days หรือไม่
    if 1 <= d <= max_days:
        return True
    else:
        return False
# --- ส่วนรับค่า input และทดสอบการใช้งาน ---
m = int(input("Enter month (m): "))
d = int(input("Enter day (d): "))
y = int(input("Enter year (y): "))
result = check_date(m, d, y)
print(f"Is valid date?: {result}")
````

5. จงเขียน function ที่รับค่าตัวเลข n เข้ามา และ return ค่าของ $n!$ (Factorial) ออกมา


````python
def factorial(n):
    result = 1
    current = 1

    # วนลูปคูณสะสมตั้งแต่ 1 ถึง n
    while current <= n:
        result *= current
        current += 1

    return result

# --- ส่วนรับค่า input และใช้งานฟังก์ชัน ---
num = int(input("Enter n: "))
print(f"{num}! = {factorial(num)}")
````

6. จงเขียน function ที่รับค่าจำนวนจริง 4 ค่า ซึ่งจะใช้แทนพิกัด (x1, y1) และ (x2, y2) โดยให้คำนวณและ return ค่าระยะทางระหว่างจุดสองจุดนี้


````python
import math

def distance(x1, y1, x2, y2):
    # คำนวณตามสูตรระยะทางระหว่างสองจุด
    d = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
    return d

# --- ส่วนรับค่า input และใช้งานฟังก์ชัน ---
x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))
x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))

result = distance(x1, y1, x2, y2)
print(f"Distance = {result:.2f}")
````

7. จงเขียน function ที่รับค่าจำนวนจริง 4 ค่า x1, y1, x2, y2 โดยที่ (x1, y1) แทนจุดศูนย์กลางของวงกลม และ (x2, y2) แทนจุดบนเส้นรอบวง เพื่อคำนวณและ return พื้นที่ของวงกลมนี้


````python
import math

def circle_area(x1, y1, x2, y2):
    # 1. คำนวณรัศมีกำลังสอง (r^2)
    r_squared = (x2 - x1)**2 + (y2 - y1)**2

    # 2. คำนวณพื้นที่วงกลม = pi * r^2
    area = math.pi * r_squared
    return area

# --- ส่วนรับค่า input และใช้งานฟังก์ชัน ---
x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))
x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))

result = circle_area(x1, y1, x2, y2)
print(f"Circle Area = {result:.2f}")
````

8. จงเขียน function ที่รับค่า m และ n จากผู้ใช้เพื่อคำนวณสูตรการจัดหมู่ $C(m, n) = \frac{m!}{n!(m - n)!}$ โดยใช้ function factorial ที่ได้เขียนไว้ก่อนหน้านี้

````python
# 1. ฟังก์ชัน Factorial จากข้อ 5 (สไตล์ while loop ของอาจารย์)
def factorial(k):
    result = 1
    current = 1
    while current <= k:
        result *= current
        current += 1
    return result

# 2. ฟังก์ชันคำนวณการจัดหมู่ C(m, n)
def combination(m, n):
    # กรณี m < n หรือค่าติดลบ จะไม่สามารถจัดหมู่ได้
    if m < n or m < 0 or n < 0:
        return 0

    # คำนวณตามสูตร C(m, n) = m! / (n! * (m - n)!)
    m_fact = factorial(m)
    n_fact = factorial(n)
    m_minus_n_fact = factorial(m - n)

    ans = m_fact // (n_fact * m_minus_n_fact)
    return ans

# --- 3. ส่วนรับค่า input จากผู้ใช้และแสดงผล ---
m = int(input("Enter m: "))
n = int(input("Enter n: "))

result = combination(m, n)
print(f"C({m}, {n}) = {result}")
````
