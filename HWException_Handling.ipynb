แบบฝึกหัดที่ 1 — Safe Divide**

เขียนฟังก์ชัน `safe_divide(a, b)` ที่คืนค่า `a / b` แต่ถ้า `b` เป็น 0 ให้คืนค่า `None` แทนที่จะทำให้โปรแกรม crash (ใช้ `try...except ZeroDivisionError`)

```python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return None

# --- ทดสอบการใช้งาน ---
print(safe_divide(10, 2))  # Output: 5.0
print(safe_divide(10, 0))  # Output: None
```

แบบฝึกหัดที่ 2 — Safe Input Number**

เขียนฟังก์ชัน `get_number(text)` ที่รับ input จากผู้ใช้และพยายามแปลงเป็น `int` หากผู้ใช้กรอกไม่ใช่ตัวเลข ให้แจ้งเตือนและวนถามใหม่จนกว่าจะได้ตัวเลขที่ถูกต้อง (ใช้ `while True` ร่วมกับ `try...except ValueError`)

```python
def get_number(text):
    while True:
        try:
            value = int(input(text))
            return value  # ถ้าแปลงเป็น int สำเร็จ จะคืนค่าและออกจากลูปทันที
        except ValueError:
            print("กรุณากรอกเฉพาะตัวเลขจำนวนเต็มเท่านั้น! ลองใหม่อีกครั้ง")

# --- ทดสอบการใช้งาน ---
num = get_number("Enter a number: ")
print(f"You entered: {num}")
```

แบบฝึกหัดที่ 3 — Dictionary Lookup**

เขียนฟังก์ชัน `get_student_grade(grades, name)` ที่รับ dictionary ของเกรดนักศึกษา และชื่อที่ต้องการค้นหา หากไม่พบชื่อนั้นใน dictionary ให้จัดการ `KeyError` และคืนค่า `"ไม่พบข้อมูลนักศึกษา"` แทน

```python
def get_student_grade(grades, name):
    try:
        return grades[name]
    except KeyError:
        return "ไม่พบข้อมูลนักศึกษา"

# --- ทดสอบการใช้งาน ---
student_grades = {"Supanat": "A", "Teerapon": "B+", "Kawinwat": "B"}

print(get_student_grade(student_grades, "Supanat"))    # Output: A
print(get_student_grade(student_grades, "Kidsakorn"))  # Output: ไม่พบข้อมูลนักศึกษา
```

แบบฝึกหัดที่ 4 — Custom Validation**

เขียนฟังก์ชัน `set_password(password)` ที่ใช้ `raise ValueError` หากรหัสผ่านสั้นกว่า 8 ตัวอักษร แล้วเรียกใช้งานด้วย `try...except` เพื่อแสดงข้อความแจ้งเตือนที่เหมาะสม



```python
def set_password(password):
    if len(password) < 8:
        raise ValueError("รหัสผ่านต้องมีความยาวอย่างน้อย 8 ตัวอักษร")
    return "ตั้งรหัสผ่านสำเร็จ!"

# --- ทดสอบการใช้งาน ---
user_pass = input("Enter new password: ")

try:
    result = set_password(user_pass)
    print(result)
except ValueError as e:
    print(f"เกิดข้อผิดพลาด: {e}")
```
